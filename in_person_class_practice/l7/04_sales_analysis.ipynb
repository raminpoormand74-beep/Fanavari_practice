````python
sales = (
    ("Ali", "Laptop", 1200),
    ("Sara", "Phone", 800),
    ("Ali", "Phone", 800),
    ("Reza", "Laptop", 1200),
    ("Sara", "Laptop", 1200),
    ("Ali", "Mouse", 50)
)
````

````python
# A-1
def calculate_customer_purchases(sales):
    """Return a dictionary containing the total purchase amount for each customer."""

    purchases = {}

    for sale in sales:
        name = sale[0]
        price = sale[2]

        if name in purchases:
            purchases[name] += price
        else:
            purchases[name] = price

    return purchases
````

````python
calculate_customer_purchases(sales)
````

````python
# A-2
def tuple_to_dict(sales):
    t_to_d = {}

    for name, product, price in sales:
        if name not in t_to_d:
            t_to_d[name] = 0

        t_to_d[name] += price

    return t_to_d
````

````python
tuple_to_dict(sales)
````

````python
# B-1
def count_product_sales(sales):
    """Return a dictionary containing the number of sales for each product."""

    product_sales = {}

    for sale in sales:
        product = sale[1]

        if product in product_sales:
            product_sales[product] += 1
        else:
            product_sales[product] = 1

    return product_sales
````

````python
count_product_sales(sales)
````

````python
# B-2
def product_information(sales):
    """Return a dictionary containing the number of sales for each product."""

    product_info = {}

    # for key, value in tuple_to_dict(sales).items():
    for name, product, number in sales:
        if product not in product_info:
            product_info[product] = 0

        product_info[product] += 1

    return product_info
````

````python
product_information(sales)
````

````python
# C-1
def calculate_total_revenue(sales):
    """Return the total revenue from all sales."""

    total_revenue = 0

    for sale in sales:
        total_revenue += sale[2]

    return total_revenue
````

````python
calculate_total_revenue(sales)
````

````python
# C-2
def total_sale(sales):
    """Return the total revenue from all sales."""

    total = 0

    for value in tuple_to_dict(sales).values():
        total += value

    return total
````
