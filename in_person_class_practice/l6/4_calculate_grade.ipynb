```python
def calculate_grade(grade):
    """
    Determine the letter grade based on a score between 0 and 100.

    Parameters:
        grade: The score to be evaluated.

    Returns:
        The corresponding letter grade.
    """
    if grade < 0 or grade > 100:
        return 'Invalid grade'

    elif 90 <= grade:
        return 'A'

    elif 80 <= grade:
        return 'B'

    elif 70 <= grade:
        return 'C'

    elif 60 <= grade:
        return 'D'

    else:
        return 'F'
```
