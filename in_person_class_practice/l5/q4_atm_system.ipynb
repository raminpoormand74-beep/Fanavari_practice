```python
while True:

    menu = ['mojoodi', 'variz', 'bardasht', 'khoroj', 'amaaliate digar']

    print(menu)

    user = input('amaliat ro entekhab kon: ')

    if user not in menu:
        print('entekhab namotabar ast')
        continue

    if user == 'mojoodi':
        print('amaliate mojoodi entekhab shod')

    elif user == 'variz':
        print('amaliate variz entekhab shod')

    elif user == 'bardasht':
        print('amaliate bardasht entekhab shod')

    elif user == 'amaaliate digar':
        continue

    elif user == 'khoroj':
        print('mamnon, khoroj ba movafaghiat anjam shod')
        break
```
