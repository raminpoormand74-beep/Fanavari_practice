```python
def check_password(password):
    """
    Check whether the password has at least 8 characters
    and contains both letters and digits.

    Parameters:
        password: The password to be checked.

    Returns:
        True if the password is valid, otherwise False.
    """

    if len(password) >= 8:

        if not password.isdigit() and not password.isalpha():

            return True

        else:
            return False

    else:
        return False
```

```python
def check_password_strength(password):
    """
    Check the strength of a password based on its length,
    letters, digits, and uppercase and lowercase characters.

    Parameters:
        password: The password to be checked.

    Returns:
        1 if the password has fewer than 8 characters.
        2 if the password has at least 8 characters but
          does not contain both letters and digits.
        3 if the password has at least 8 characters and
          contains both letters and digits.
        4 if the password contains letters, digits,
          uppercase and lowercase characters.
    """

    if len(password) < 8:
        return 1

    if not password.isdigit() and not password.isalpha():

        if not password.islower() and not password.isupper():
            return 4

        else:
            return 3

    else:
        return 2
```
