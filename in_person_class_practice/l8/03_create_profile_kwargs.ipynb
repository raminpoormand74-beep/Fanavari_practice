Tabei be name `create_profile` benevisid ke do voroodiye ejbari be namhaye `name` va `age` daryaft konad.

Elave bar in, tabe bayad betavanad har tedad etelaate ezafi ra ba estefade az `**kwargs` daryaft konad.

Baraye mesal:

* City
* Job
* Email
* Va har etelaate ezafiye digari

Tabe bayad karhaye zir ra anjam dahad:

* Nam va sene karbar ra namayesh dahad.
* Tamam etelaate ezafi daryaft shode ra namayesh dahad.
* Tedade etelaate ezafi ra chap konad.
* Agar `city` vojood dasht, shahr ra be sorate jodagane namayesh dahad.
* Agar `email` vojood nadasht, ebarat zir ra namayesh dahad:

```text
email not provided
```

Dar nahayat tamam etelaate karbar ra dakhele yek dictionarye jadid gharar dade va an ra `return` konid.


````python
def Create_profile(name, age, **kwargs):
    """
    Create a user profile with name, age, and additional information.

    Args:
        name (str): User's name.
        age (int): User's age.
        **kwargs: Additional user information such as city, job, and email.

    Returns:
        dict: A dictionary containing all user information.
    """
    
    print(f'name: {name}')
    print(f'age: {age}')
    print()
    print('etelaat digar: ')
    
    for key, value in kwargs.items():
        print(f'{key} : {value}')

    print()
    print(f'tedad etelaat ezafi: {len(kwargs)}')

    if 'city' in kwargs:
        print(f'shahr: {kwargs["city"]}')
        
    if 'email' not in kwargs:
        print('Email not provided')

    profile = {
        'name': name,
        'age': age
    }

    profile.update(kwargs)

    return profile
````

````python
Create_profile('sanaz', 20, city='esfahan', job='programmer', email='sanaz20@gmail.com')
````
