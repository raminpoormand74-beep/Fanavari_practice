```python
def products_checker(products, code):
    """
    Returns the price of a product based on its code.

    Args:
        products (list): A list of dictionaries containing product information.
        code (str): The code of the product to search for.

    Returns:
        int or float: The price of the product if the code is found.
    """

    for product in products:

        if product['code'] == code:

            return product['price']
```

```python
def products_checker(products, code):
    """
    Returns the complete information of a product based on its code.

    Args:
        products (list): A list of dictionaries containing product information.
        code (str): The code of the product to search for.

    Returns:
        tuple: A tuple containing the product code, name, and price.
    """

    for product in products:

        if product['code'] == code:

            return (product['code'], product['name'], product['price'])
```
