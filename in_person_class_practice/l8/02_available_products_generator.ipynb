Tabei be name `available_products` benevisid ke yek dictionary az mahsoolat ra be onvane voroodi daryaft konad.

Baraye mesal:

```python
products = {
    "laptop": 3,
    "phone": 0,
    "tablet": 5,
    "mouse": 0,
    "keyboard": 2
}
```

Sepas faghat name mahsoolati ra ke mojoodiye anha bishtar az sefr ast, ba estefade az `yield` yeki yeki tolid konid.

Dar edame yek Generator besazid va mahsoolate mojood ra ba estefade az `next()` dane-dane daryaft konid.


````python
products = {
    "laptop": 3,
    "phone": 0,
    "tablet": 5,
    "mouse": 0,
    "keyboard": 2
}
````

````python
def available_products(products):
    for key, value in products.items():
        if value > 0:
            yield key

Generator = available_products(products)
    
````

````python
print(next(Generator))
````

````python
print(next(Generator))
````

````python
print(next(Generator))
````
