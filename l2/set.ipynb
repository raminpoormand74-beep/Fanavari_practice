```python
name = {53, 12,9,56,4}
print(len(name))#tool variable ro barmigardoone
name.add(11)#yak onsor ezafe mikone
print(name)
name.update([14])#yek onsor ro be akhar set ezafe mikone
print(name)
name.remove(11)#pak kardan yek onsor
print(name)
name.discard(2)#namayesh bedon error ya khata hata vaghti onsori ro mojood nadare
print(name)
print(name.pop())#be tor tasadofi yek onsor ra barmigardoone
print(sorted(name))#moratab sazi
name.clear()#pak kardan ham
print(name)
```
