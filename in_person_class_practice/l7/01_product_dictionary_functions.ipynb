````python
products = {
'laptab': 1200,
'phone': 800,
'tablet': 500,
'headphone': 150,
'mouse': 50
}
````

````python
# A-1
def find_max_price(products):
    """
    Find and return the highest price among all products.

    Args:
        products (dict): A dictionary containing product names and prices.

    Returns:
        int: The highest product price.
    """

    max_price = 0

    for price in products.values():
        if price > max_price:
            max_price = price

    return max_price
````

````python
find_max_price(products)
````

````python
# A-2
def find_max_price_name(products):
    """
    Find and return the name and price of the most expensive product.

    Args:
        products (dict): A dictionary containing product names and prices.

    Returns:
        tuple: The name and price of the most expensive product.
    """

    max_price = 0
    max_price_name = 0

    for key, value in products.items():
        if value > max_price:
            max_price = value
            max_price_name = key

    return max_price_name, max_price
````

````python
find_max_price_name(products)
````

````python
# A-3

def find_min_price(products):
    """
    Find and return the name and price of the cheapest product.

    Args:
        products (dict): A dictionary containing product names and prices.

    Returns:
        tuple: The name and price of the cheapest product.
    """

    min_price = list(products.values())[0]
    min_price_name = list(products.keys())[0]

    for name, price in products.items():
        if price < min_price:
            min_price = price
            min_price_name = name

    return min_price_name, min_price
````

````python
find_min_price(products)
````

````python
# A-4
def calculate_total_price(products):
    """
    Calculate and return the total price of all products.

    Args:
        products (dict): A dictionary containing product names and prices.

    Returns:
        int: The total price of all products.
    """

    total = 0
    for i in products.values():
        total = total + i
    return total
````

````python
calculate_total_price(products)
````

````python
# A-5-1
def my_len(products):
    """
    Count and return the number of products.

    Args:
        products (dict): A dictionary containing product names and prices.

    Returns:
        int: The number of products.
    """

    len_products = 0
    for i in products:
        len_products = len_products + 1
    return len_products
````

````python
my_len(products)
````

````python
# A-5-2
def calculate_average_price(products):
    """
    Calculate and return the average price of all products.

    Args:
        products (dict): A dictionary containing product names and prices.

    Returns:
        float: The average price of all products.
    """

    average_price = calculate_total_price(products) / my_len(products)

    return average_price
````

````python
calculate_average_price(products)
````
