```python
# A
def add_user(users, username, age, city):
    """
    Adds a new user to the users list.

    Args:
        users (list): A list of dictionaries containing user information.
        username (str): The username of the new user.
        age (int): The age of the new user.
        city (str): The city of the new user.

    Returns:
        list: The updated list containing the new user.
    """

    user_info = {
        'username': username,
        'age': age,
        'city': city,
        'active': True
    }

    users.append(user_info)

    return users
```

```python
add_user([
 {
 "username": "ali",
 "age": 25,
 "city": "Tehran",
 "active": True
 },
 {
 "username": "sara",
 "age": 17,
 "city": "Tabriz",
 "active": True
} 
],'alireza',67,'mashhad')
```

```python
# B
def find_user(users, username):
    """
    Finds a user by their username.

    Args:
        users (list): A list of dictionaries containing user information.
        username (str): The username of the user to search for.

    Returns:
        dict or None: The user's dictionary if found, otherwise None.
    """

    for user in users:

        if user['username'] == username:
            return user

    return None
```

```python
# C
def check_access(users, username):
    """
    Checks whether a user has access based on their age and active status.

    Args:
        users (list): A list of dictionaries containing user information.
        username (str): The username of the user to check.

    Returns:
        bool: True if the user is at least 18 years old and active,
        otherwise False.
    """

    for user in users:

        if user['username'] == username:

            if user['age'] >= 18 and user['active'] == True:
                return True

            else:
                return False

    return False
```

```python
# D
def get_adult_users(users):
    """
    Returns a list containing users who are over 18 years old.

    Args:
        users (list): A list of dictionaries containing user information.

    Returns:
        list: A list containing users whose age is greater than 18.
    """

    adult_users = []

    for user in users:

        if user['age'] > 18:
            adult_users.append(user)

    return adult_users
```
