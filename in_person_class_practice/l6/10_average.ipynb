```python
def average_grade(list_grade):
    """
    Calculates the average of the grades in a list.

    Args:
        list_grade (list): A list of numeric grades.

    Returns:
        float: The average of the grades.
    """
    count = 0 
    for i in list_grade:
        count += i
    count = count / len(list_grade)
    return count
```

```python
average_grade([88,42,450,74])
```
