```python
def calculator(number_1, number_2, operation):
    """
    Performs addition, subtraction, multiplication, or division
    on two given numbers.

    Args:
        number_1 (int or float): The first number.
        number_2 (int or float): The second number.
        operation (str): The operation to perform.

    Returns:
        int or float: The result of the selected operation.
    """

    if operation == '+':
        return number_1 + number_2

    elif operation == '-':
        return number_1 - number_2

    elif operation == '*':
        return number_1 * number_2

    elif operation == '/':
        return number_1 / number_2
```

```python
calculator(12.5,3,'+')
```

```python
calculator(12,3,'-')
```

```python
calculator(12,3,'*')
```

```python
calculator(12,3,'/')
```

```python
def calculator(number_1, number_2, operation):
    """
    Performs a mathematical operation on two numbers.

    Args:
        number_1 (int or float): The first number.
        number_2 (int or float): The second number.
        operation (str): The mathematical operation (+, -, *, /).

    Returns:
        int or float: The result of the operation.
        None: If the operation is invalid.
    """

    if operation == '+':
        return number_1 + number_2

    elif operation == '-':
        return number_1 - number_2

    elif operation == '*':
        return number_1 * number_2

    elif operation == '/':
        return number_1 / number_2

    else:
        return None
```
