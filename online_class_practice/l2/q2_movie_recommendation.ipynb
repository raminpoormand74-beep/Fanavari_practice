```python
sen = int(input('sen khod ra vared kon: '))
janre = input('janre morede alaghe ra vared kon: ')
hadaghal_emtiyaz = float(input('hadaghal emtiyaz ra vared kon: '))

if janre == 'elmi' and sen >= 13 and hadaghal_emtiyaz <= 8.7:
    print('film pishnahadi: Interstellar')

elif janre == 'animation' and sen >= 0 and hadaghal_emtiyaz <= 8.3:
    print('film pishnahadi: Toy Story')

elif janre == 'action' and sen >= 15 and hadaghal_emtiyaz <= 9:
    print('film pishnahadi: The Dark Knight')

elif janre == 'tarsnak' and sen >= 17 and hadaghal_emtiyaz <= 7.5:
    print('film pishnahadi: The Conjuring')

else:
    print('film monasebi peyda nashod')
```
