```python
def failed_score(list_score):
    """
    Creates a new list containing passing scores.

    Args:
        list_score (list): A list of scores.

    Returns:
        str: A message containing the list of passing scores.
    """

    passed_list = []

    for i in list_score:

        if i >= 10:

            passed_list.append(i)

    return f'list ghabooli ha: {passed_list}'
```

```python
failed_score([18, 7, 13, 9, 20, 5])
```

```python
def failed_score(list_score):
    """
    Creates a new list containing failed scores.

    Args:
        list_score (list): A list of scores.

    Returns:
        str: A message containing the list of failed scores.
    """

    failed_list = []

    for i in list_score:

        if i < 10:

            failed_list.append(i)

    return f'list mardodi ha: {failed_list}'
```

```python
def failed_score(list_score):
    """
    Counts the number of passing scores in a list.

    Args:
        list_score (list): A list of scores.

    Returns:
        str: A message containing the number of passing scores.
    """

    tedad_ghabooli = 0

    for i in list_score:

        if i >= 10:

            tedad_ghabooli += 1

    return f'tedad ghabooli ha: {tedad_ghabooli}'
```
