`````python
# A
def find_highest_salary(employees):
    """Return the name and salary of the employee with the highest salary."""

    most_salary = 0
    name_most_salary = ''

    for value in employees.values():
        if value['salary'] > most_salary:
            most_salary = value['salary']
            name_most_salary = value['name']

    return f'balatarin hoghogh ---> {name_most_salary}: {most_salary}'
`````

`````python
find_highest_salary(employees)
`````

`````python
# B
def get_lowest_salary_employee(employees):
    """Return the name and salary of the employee with the lowest salary."""

    lowest_salary = list(employees.values())[0]['salary']
    name_lowest_salary = list(employees.values())[0]['name']

    for value in employees.values():
        if value['salary'] < lowest_salary:
            lowest_salary = value['salary']
            name_lowest_salary = value['name']

    return f'kamtarin hoghogh ---> {name_lowest_salary}: {lowest_salary}'
`````

`````python
get_lowest_salary_employee(employees)
`````

`````python
# C
def find_employees_over_3000(employees):
    """Returns a list of employees whose salary is more than 3000."""
    
    employees_list = []

    for key, value in employees.items():
        if value["salary"] > 3000:
            employees_list.append(value["name"])

    return employees_list
`````

`````python
find_employees_over_3000(employees)
`````

`````python
# D
def find_employees_by_salary(employees, salary):
    """Returns a list of employees whose salary is more than the given salary."""

    employees_list = []

    for key, value in employees.items():
        if value["salary"] > salary:
            employees_list.append(value["name"])

    return employees_list
`````

`````python
find_employees_by_salary(employees, 2500)
`````

`````python
# E
def calculate_average_salary(employees):
    """Returns the average salary of all employees."""

    total_salary = 0

    for key, value in employees.items():
        total_salary += value["salary"]

    average = total_salary / len(employees)

    return average
`````

`````python
calculate_average_salary(employees)
`````
