```python
def find_product(products, code):
    """
    Finds a product by its code.

    Args:
        products (list): A list of dictionaries containing product information.
        code (str): The code of the product to find.

    Returns:
        dict or None: The matching product if found, otherwise None.
    """

    for product in products:

        if product['code'] == code:

            return product

    return None
```

```python
cart = []

def add_to_cart(products, cart, code):
    """
    Adds a product to the cart if the product exists and is in stock.

    Args:
        products (list): A list of dictionaries containing product information.
        cart (list): A list containing the products added to the cart.
        code (str): The code of the product to add to the cart.

    Returns:
        list: The updated cart containing the selected product.
    """

    for product in products:

        if product['code'] == code:

            if product['stock'] > 0:

                cart.append(product)

    return cart
```

```python
products = [
    {"code": "p1", "name": "Keyboard", "price": 50, "stock": 4},
    {"code": "p2", "name": "Mouse", "price": 30, "stock": 7},
    {"code": "p3", "name": "Monitor", "price": 250, "stock": 2},
    {"code": "p4", "name": "Headphone", "price": 80, "stock": 0}
]

add_to_cart(products, cart, 'p2')
```

```python
cart = []


def add_to_cart(products, cart, code):
    """
    Adds an in-stock product to the cart and decreases its stock by one.

    Args:
        products (list): A list of dictionaries containing product information.
        cart (list): A list containing the products added to the cart.
        code (str): The code of the product to add to the cart.

    Returns:
        list: The updated products list.
    """

    for product in products:

        if product['code'] == code:

            if product['stock'] > 0:

                product['stock'] -= 1

                cart.append(product)

    return products
```

```python
def add_to_cart(products, code):
    """
    Decreases the stock of a product by one if it exists and is in stock.

    Args:
        products (list): A list of dictionaries containing product information.
        code (str): The code of the product whose stock should be decreased.

    Returns:
        list: The updated products list.
    """

    for product in products:
        if product['code'] == code:
            if product['stock'] > 0:
                product['stock'] -= 1

    return products
```

```python
cart = []


def add_to_cart(products, code):
    """
    Adds an in-stock product to the cart and decreases its stock by one.

    Args:
        products (list): A list of dictionaries containing product information.
        code (str): The code of the product to add to the cart.

    Returns:
        tuple: A tuple containing the updated products list and cart.
    """

    for product in products:
        if product['code'] == code:
            if product['stock'] > 0:
                cart.append(product)
                product['stock'] -= 1

    return products, cart
```
