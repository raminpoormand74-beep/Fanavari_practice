```python
# B
def calculate_age(birth_year, date_type):
    """
    Calculate age based on the birth year and date type.

    Parameters:
        birth_year: The birth year.
        date_type: The calendar type, either 'miladi' or 'shamsi'.
    """
    if date_type == 'miladi':
        year = 2026
    else:
        year = 1405

    return f'sale tavalod shoma hast: {birth_year} va sen shoma hast: {year - birth_year}'
```

```python
calculate_age(1996,'miladi')
```

```python
calculate_age(1365,'shamsi')
```

```python
# C
def calculate_age(birth_year, date_type='miladi'):
    """
    Calculate age based on the birth year and calendar type.

    Parameters:
        birth_year: The birth year.
        date_type: The calendar type, either 'miladi' or 'shamsi'.
                   The default value is 'miladi'.
    """
    if date_type == 'miladi':
        sale_tavalod = 2026 - birth_year

    elif date_type == 'shamsi':
        sale_tavalod = 1405 - birth_year

    return f'sale tavalod shoma hast: {birth_year} va sen shoma hast: {sale_tavalod}'
```

```python
calculate_age(1989)
```

```python
calculate_age(1383,'shamsi')
```

```python
# D
def calculate_age(birth_year):
    """
    Calculate age based on the birth year.

    Parameters:
        birth_year: The birth year in either the Shamsi or Gregorian calendar.
                   Shamsi years should be between 1300 and 1405,
                   and Gregorian years should be between 1900 and 2026.
    """
    if 1300 <= birth_year <= 1405:
        sale_tavalod = 1405 - birth_year

    elif 1900 <= birth_year <= 2026:
        sale_tavalod = 2026 - birth_year

    return f'sale tavalod shoma hast: {birth_year} va sen shoma hast: {sale_tavalod}'
```

```python
calculate_age(1385)
```

```python
calculate_age(1984)
```
