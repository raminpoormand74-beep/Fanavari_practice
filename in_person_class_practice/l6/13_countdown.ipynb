```python
def countdown(number):
    """
    Prints a countdown from the given number to zero.

    Args:
        number (int): The starting number of the countdown.

    Returns:
        None: Prints the numbers from the given number down to 0.
    """

    for i in range(number, -1, -1):
        print(i)
```

```python
countdown(12)
```

```python
# inja bedon estefade az print vali baraye namayesh az yek list estefade kardam va return list ro gereftam
def countdown(number):
    """
    Creates a countdown list from the given number to zero.

    Args:
        number (int): The starting number of the countdown.

    Returns:
        list: A list containing the numbers from the given number down to zero.
    """

    countdown_list = []

    for i in range(number, -1, -1):
        countdown_list.append(i)

    return countdown_list
```

```python
countdown(9)
```
