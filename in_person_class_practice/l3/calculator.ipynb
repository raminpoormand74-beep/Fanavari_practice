```python
"""in baraye inke betoonim adad ashari ham bedim mesl 8.5 
vali tashkhis horoof ro nadare yani agar horoof bejaye adad bedim error mide 
inja mikhastam code ro tori benevisam ke agar karbar horoof vared kard msg lotfan adad vared konid bede
vali chon dar ebteda input haro float kardam dige nemishod az isdigit() estefade konam
agar estefade mikardam dige nemishod adad ashari dad"""

user_1 = float(input('adad aval ra vared konid'))
op = input('Enter your operation')
user_2 = float(input('adad dovom ra vared konid'))
print(user_1,op,user_2)
calculator = (user_1,user_2)
if calculator and op :
        if op == '+':
            bealave = user_1+user_2
            if bealave.is_integer():
                print(int(bealave))
            else :
                print(bealave)
        elif op == '*':
            zarb = user_1*user_2
            if zarb.is_integer():
                print(int(zarb))
            else :
                print(zarb)
        elif op == '/':
            taghsim = user_1/user_2
            if taghsim.is_integer():
                print(int(taghsim))
            else :
                print(taghsim)
        elif op == '//':
            taghsim_sahih = user_1//user_2
            if taghsim_sahih.is_integer():
                print(int(taghsim_sahih))
            else:
                print(taghsim_sahih)
        elif op == '-':
            tafrigh = user_1-user_2
            if tafrigh.is_integer():
                print(int(tafrigh))
            else:
                print(tafrigh)
        elif op != '-' and '+' and '*' and '/' and '//':
            print('lotfan amalgar(+,-,/,*,//)ra vared konid')
        else:
            print('----lotfan faghat adad vared konid')

            
```
