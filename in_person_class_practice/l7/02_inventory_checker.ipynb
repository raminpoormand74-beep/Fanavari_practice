````python
inventory = {
    "apple": 20,
    "banana": 5,
    "orange": 0,
    "milk": 12,
    "bread": 0
}
````

````python
def check_inventory(inventory):
    in_stock = []
    none_stock = []
    for key, value in inventory.items():
        if value > 0:
            in_stock.append(key)
        else:
            none_stock.append(key)

    return f'mahsoolat mojood: {in_stock} va mahsoolat namojood: {none_stock}'
````

````python
check_inventory(inventory)
````
