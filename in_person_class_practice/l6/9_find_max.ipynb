```python
def find_max_number(list_numbers):
    """
    Finds and returns the largest number in a list.

    Args:
        list_numbers (list): A list of numbers.

    Returns:
        str: A message containing the largest number in the list.
    """
    max_number = list_numbers[0]

    for number in list_numbers:
        if number > max_number:
            max_number = number
    
    return f'bozorgtarin adad list: {max_number} mibashad'
```
