```python
def check_number(number):
    """
    Check whether the given number is even or odd.

    Parameters:
        number: The number to be checked.
    """
    if number % 2 == 0:
        return f'{number} is even'

    else: 
        return f'{number} is odd'
```

```python
check_number(18)
```

```python
check_number(13)
```

```python
def check_number(number):
    """
    Check whether the given number is even.

    Parameters:
        number: The number to be checked.

    Returns:
        True if the number is even, otherwise False.
    """
    if number % 2 == 0:
        return True

    return False
```

```python
check_number(16)
```

```python
check_number(13)
```

```python
def check_number(number):
    """
    Check whether the given number is positive, negative, or zero.

    Parameters:
        number: The number to be checked.

    Returns:
        A message indicating whether the number is positive, negative, or zero.
    """
    if number > 0:
        return 'number is mosbat'

    elif number < 0:
        return 'number is manfi'

    else:
        return 'number is zero'
```

```python
check_number(0)
```

```python
check_number(14)
```

```python
check_number(-8)
```
