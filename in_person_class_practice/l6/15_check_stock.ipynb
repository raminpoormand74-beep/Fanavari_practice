```python
def check_stock(products, name_product):
    """
    Checks whether a product is available in stock.

    Args:
        products (dict): A dictionary containing product names and their stock quantities.
        name_product (str): The name of the product to check.

    Returns:
        bool: True if the product exists and its stock is greater than zero,
        otherwise False.
    """

    if name_product in products:

        if products[name_product] > 0:

            return True

        else:

            return False

    return False
```

```python
def check_stock(products, name_product):
    """
    Checks whether a product exists and is available in stock.

    Args:
        products (dict): A dictionary containing product names and their stock quantities.
        name_product (str): The name of the product to check.

    Returns:
        bool: True if the product exists and its stock is not zero,
        otherwise False.
    """

    for key, value in products.items():
        if key == name_product and value != 0:
            return True
        elif key == name_product and value == 0:
            return False

    return False
```

```python
products = {
    "iphone": 5,
    "macbook": 2,
    "airpods": 0
}

check_stock(products, 'samsung')
```
