```python
def name_cleaner(first_name, last_name):
    """
    Clean and format the first and last name.

    Parameters:
        first_name: The person's first name.
        last_name: The person's last name.

    Returns:
        A formatted string containing the cleaned first and last name.
    """

    first_name = first_name.strip().title()
    last_name = last_name.strip().title()

    return f'etelat vared shode: {first_name} {last_name}'
```
