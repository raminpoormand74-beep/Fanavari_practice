```python
def check_age(age):
    """
    Check whether the given age is allowed to access.

    Parameters:
        age: The age to be checked.

    Returns:
        'Access Denied' if the age is less than 18,
        otherwise 'Welcome'.
    """
    if age < 18:
        return 'Access Denied'

    else:
        return 'Welcome'
```

```python
check_age(22)
```

```python
check_age(16)
```
