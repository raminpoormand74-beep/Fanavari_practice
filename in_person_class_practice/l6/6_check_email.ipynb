```python
def check_email(email):
    """
    Check whether an email address is valid based on specific conditions.

    Parameters:
        email: The email address to be checked.

    Returns:
        True if the email contains no spaces, includes '@' and '.com',
        otherwise False.
    """
    if ' ' in email:
        return False
    if '@' not in email:
        return False
    if '.com' not in email:
        return False
    return True
```

```python
check_email('ali@gmail.com')
```

```python
check_email('aligamil.com')
```

```python
check_email('ali@gmail.')
```
