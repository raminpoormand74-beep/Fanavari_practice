```python
def multiplication_table(number):
    """
    Prints the multiplication table of a given number from 0 to 9.

    Args:
        number (int): The number for which the multiplication table is generated.

    Returns:
        None: Prints the multiplication table.
    """

    for i in range(10):

        print(f'{number} × {i} = ', i * number)
```

```python
multiplication_table(5)
```
