```python
def Menu_app():
    """
    Receives food orders from the user and returns all valid orders.

    The user can enter food names from the menu repeatedly.
    Entering 'order' ends the ordering process.

    Returns:
        list: A list containing all valid food orders.
    """

    foods = ['pizza', 'burger', 'pasta', 'lazania']

    orders = []

    while True:

        order = input('esm ghaza ro vared kon: ')

        if order == 'order':
            break

        if order in foods:
            orders.append(order)

    return orders
```
