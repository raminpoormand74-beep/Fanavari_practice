### 1. Definition of the Function

Tabei be name `even_numbers` benevisid ke yek adad be name `n` ra be onvane voroodi daryaft konad.

### 2. Using `yield`

Sepas ba estefade az dastoor-e `yield`، adad-e zoj ra az `2` ta `n` yeki yeki tolid konad.

### 3. Create a Generator

Dar edame، yek Generator besazid va an ra dakhele yek moteghayer gharar dahid.

### 4. Using `next()`

Ba estefade az `next()` se meghdar-e aval ra namayesh dahid.


```python
def even_numbers(n):

    """Generate even numbers from 2 up to n."""
    
    for number in range(2, n, 2):
        yield number
```

```python
number = even_numbers(17)
```

```python
print(next(number))
print(next(number))
print(next(number))
```
